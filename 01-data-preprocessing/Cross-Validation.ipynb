# Cross Validation

Cross validation is a technique used to check how well a machine learning model performs on unseen data while preventing overfitting.

**It works by:**
1. Splitting the dataset into several parts.
2. Training the model on some parts and testing it on the remaining part.
3. Repeating this sampling process multiple times by choosing different parts of the dataset.
4. Averaging the results from each validation step to get the final performance.


## Validation Techniques

### 1. Holdout Validation

In this validation, the dataset is split into two parts:
1. Training set
2. Test set

The model learns on the training set and is evaluated on the test set.

**Common splits:** 70-30, 80-20, and 75-25 depending on the dataset size and problem. It is simple and quick to apply.


### Why do we need it?

Suppose we have a dataset of 100 students. If we train the model on all 100 students and then test it on the same 100 students, the model will perform well and give 100% accuracy because the model sees the same data for both training and testing. The model does not learn patterns but memorizes the answers. This is called **overfitting**.

**Solution:**
- Train model on 80 students
- Test model on 20 students

Now the model sees new data (unseen) and will not give 100% accuracy because this time it learns patterns instead of memorizing answers.
